# Modelos de Regressão

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, mean_absolute_percentage_error
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from xgboost import XGBRegressor
```

```python
df = pd.read_csv('kc_house_data.csv')
df.head()
```

```python
df.shape
```

```python
df = df[['id','date','price','bedrooms','bathrooms','sqft_living','sqft_lot','floors','waterfront']]
df.head()
```

```python
columns_to_drop = ['id','date']
df = df.drop(columns=columns_to_drop)
df.head()
```

```python
rows,cols = df.shape
print(f"O dataset possui {rows} linhas e {cols} colunas.")
```

```python
x = df.drop(columns='price')
y = df['price']
```

```python
x.head(),y.head()   
```

```python
scaler = StandardScaler()
x = scaler.fit_transform(x)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
len(y_train),len(y_test)
```

## Regressão Linear

```python
lr_model = LinearRegression()
lr_model.fit(x_train,y_train)

y_pred = lr_model.predict(x_test)
```

```python
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse) 
mape = mean_absolute_percentage_error(y_test, y_pred)

print(f"MSE : {mse} \nRMSE: {rmse} \nMAPE: {mape}")
```

## K-NN Regressor

**K=7**

```python
knr_model = KNeighborsRegressor(n_neighbors=7, metric='euclidean')
knr_model.fit(x_train,y_train)

y_pred_knr = knr_model.predict(x_test)
```

```python
mape = mean_absolute_percentage_error(y_test, y_pred_knr)
print(f"MAPE: {mape}")
```

**K=11**

```python
knr_model = KNeighborsRegressor(n_neighbors=11, metric='euclidean')
knr_model.fit(x_train,y_train)

y_pred_knr = knr_model.predict(x_test)
```

```python
mape = mean_absolute_percentage_error(y_test, y_pred_knr)
print(f"MAPE: {mape}")
```

## SVM Support Vector Machines for Regression

```python
svr_model = SVR()
svr_model.fit(x_train,y_train)
y_pred_svr = svr_model.predict(x_test)
```

```python
mape = mean_absolute_percentage_error(y_test, y_pred_svr)
print(f"MAPE: {mape}")
```

**Testing other kernel**

```python
svr_model = SVR(kernel='linear', C=100)
svr_model.fit(x_train,y_train)
y_pred_svr = svr_model.predict(x_test)
```

```python
mape = mean_absolute_percentage_error(y_test, y_pred_svr)
print(f"MAPE: {mape}")
```

## Árvore de Regressão

```python
dcr_model = DecisionTreeRegressor()
dcr_model.fit(x_train,y_train)
y_pred_dcr = dcr_model.predict(x_test)
```

```python
mape = mean_absolute_percentage_error(y_test, y_pred_dcr)
print(f"MAPE: {mape}")
```

## XGBoost


```python
xgb_model = XGBRegressor()
xgb_model.fit(x_train,y_train)
y_pred_xgb = xgb_model.predict(x_test)
```

```python
mape = mean_absolute_percentage_error(y_test, y_pred_xgb)
print(f"MAPE: {mape}")
```
