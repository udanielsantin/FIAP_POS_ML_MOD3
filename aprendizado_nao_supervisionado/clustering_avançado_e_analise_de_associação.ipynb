### Apriori Movies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from apyori import apriori
```

```python
movie_data = pd.read_csv('https://raw.githubusercontent.com/ahirtonlopes/Unsupervised-Learning/main/movie_dataset.csv', header=None)

movie_data.shape
```

```python
movie_data.head()
```

```python
records = []
for i in range(0, movie_data.shape[0]):
    records.append([str(movie_data.values[i,j]) for j in range(0,20)])
```

```python
association_rules = apriori(records, min_support = 0.004, min_confidence = 0.20, min_lift = 3, min_length = 2)
association_results = list(association_rules)
```

```python
print(len(association_results))
```

```python
print(association_results[0])
```

```python
results = []
for item in association_results:
    pair = item[0]
    items = [x for x in pair]
    
    value0 = str(items[0])
    value1 = str(items[1])

    value2 = str(item[1])[:7]

    value3 = str(item[2][0][2])[:7]
    value4 = str(item[2][0][3])[:7]

    rows = (value0, value1, value2, value3, value4)
    results.append(rows)

labels = ['Titulo 1', 'Titulo 2', 'Support', 'Confidence', 'Lift']
movie_suggestion = pd.DataFrame.from_records(results, columns=labels)

print(movie_suggestion)
```
